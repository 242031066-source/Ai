```python
import networkx as nx
import matplotlib.pyplot as plt
from collections import deque

graph = {
    '2': ['4', '0'],
    '4': ['3', '1'],
    '0': ['5'],
    '3': ['5'],
    '1': ['5'],
    '5': []
}

def BFS(start, target):

    queue = deque([[start]])
    visited = []
    steps = []

    while queue:

        path = queue.popleft()
        node = path[-1]

        if node in visited:
            continue

        visited.append(node)
        steps.append((node, visited.copy()))

        if node == target:
            print("Path:", path)
            print("Visited Nodes:", visited)
            break

        for neighbor in graph[node]:

            if neighbor not in visited:
                queue.append(path + [neighbor])

    G = nx.Graph()

    for node, neighbors in graph.items():
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

    pos = {
        '2': (0, 3),
        '4': (-1, 2),
        '0': (1, 2),
        '3': (-1.5, 1),
        '1': (-0.5, 1),
        '5': (1, 1)
    }

    for node, done in steps:

        plt.figure(figsize=(8, 5))

        colors = [
            'orange' if n == node
            else 'lightgreen' if n in done
            else 'skyblue'
            for n in G.nodes()
        ]

        nx.draw(
            G, pos,
            with_labels=True,
            node_color=colors,
            node_size=2200,
            font_size=15,
            font_weight='bold'
        )

        plt.title(f"BFS - ID 242031066 | Visiting: {node}")
        plt.show()


BFS('2', '5')
```

```python
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    '2': ['4', '0'],
    '4': ['3', '1'],
    '0': ['5'],
    '3': ['5'],
    '1': ['5'],
    '5': []
}

def IDS(start, target, max_depth):

    G = nx.Graph()

    for node, neighbors in graph.items():
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

    pos = {
        '2': (0, 3),
        '4': (-1, 2),
        '0': (1, 2),
        '3': (-1.5, 1),
        '1': (-0.5, 1),
        '5': (1, 1)
    }

    for limit in range(max_depth + 1):

        print("\nDepth Limit:", limit)

        stack = [[start]]
        visited = []
        steps = []
        found = False

        while stack:

            path = stack.pop()
            node = path[-1]
            depth = len(path) - 1

            if node in visited:
                continue

            visited.append(node)
            steps.append((node, visited.copy()))

            if node == target:
                print("Path:", path)
                print("Visited Nodes:", visited)
                found = True
                break

            if depth < limit:

                for neighbor in reversed(graph[node]):

                    if neighbor not in visited:
                        stack.append(path + [neighbor])

        for node, done in steps:

            plt.figure(figsize=(8, 5))

            colors = [
                'orange' if n == node
                else 'lightgreen' if n in done
                else 'skyblue'
                for n in G.nodes()
            ]

            nx.draw(
                G, pos,
                with_labels=True,
                node_color=colors,
                node_size=2200,
                font_size=15,
                font_weight='bold'
            )

            plt.title(f"IDS - ID 242031066 | Depth: {limit} | Visiting: {node}")
            plt.show()

        if found:
            return

    print("Target not found")


IDS('2', '5', 5)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import heapq

graph = {
    '2': [('4', 1), ('0', 4)],
    '4': [('3', 2), ('1', 5)],
    '0': [('5', 3)],
    '3': [('5', 2)],
    '1': [('5', 1)],
    '5': []
}

def UCS(start, target):

    pq = [(0, [start])]
    visited = []
    steps = []

    while pq:

        cost, path = heapq.heappop(pq)
        node = path[-1]

        if node in visited:
            continue

        visited.append(node)
        steps.append((node, visited.copy()))

        if node == target:
            print("Path:", path)
            print("Total Cost:", cost)
            print("Visited Nodes:", visited)
            break

        for neighbor, weight in graph[node]:

            if neighbor not in visited:
                heapq.heappush(
                    pq,
                    (cost + weight, path + [neighbor])
                )

    G = nx.Graph()

    for node, neighbors in graph.items():
        for neighbor, weight in neighbors:
            G.add_edge(node, neighbor, weight=weight)

    pos = {
        '2': (0, 3),
        '4': (-1, 2),
        '0': (1, 2),
        '3': (-1.5, 1),
        '1': (-0.5, 1),
        '5': (1, 1)
    }

    edge_labels = nx.get_edge_attributes(G, 'weight')

    for node, done in steps:

        plt.figure(figsize=(8, 5))

        colors = [
            'orange' if n == node
            else 'lightgreen' if n in done
            else 'skyblue'
            for n in G.nodes()
        ]

        nx.draw(
            G, pos,
            with_labels=True,
            node_color=colors,
            node_size=2200,
            font_size=15,
            font_weight='bold'
        )

        nx.draw_networkx_edge_labels(
            G, pos,
            edge_labels=edge_labels,
            font_size=12
        )

        plt.title(f"UCS - ID 242031066 | Visiting: {node}")
        plt.show()


UCS('2', '5')
```
